```python
# Load CSV into DataFrame
df = spark.read.csv("/Volumes/navneet_data/default/navneet_bda_data/student_scores(1).csv", header=True,
inferSchema=True)
df.show(5)
```

```python
# Filter students with math_score > 80
high_math = df.filter(df.math_score > 80)
high_math.show()
```

```python
# Select name and reading_score
df.select("name", "reading_score").show(5)
```

```python
# Average scores by gender
avg_scores = df.groupBy("gender").agg(
 {"math_score": "avg", "reading_score": "avg", "writing_score": "avg"}
)
avg_scores.show()
```

```python
# Create another DataFrame with student hobbies
hobbies = [
 (1, "Football"),
 (2, "Chess"),
 (3, "Music"),
 (4, "Dance")
]
hobby_df = spark.createDataFrame(hobbies, ["student_id", "hobby"])
# Join with main DataFrame
joined_df = df.join(hobby_df, "student_id", "left")
joined_df.show()
```

```python
# Top 5 students by writing score
df.orderBy(df.writing_score.desc()).show(5)
```
